## Step 1: Load the Dataset
The sales dataset is loaded using Pandas.  
The dataset file is stored in the same folder as this notebook.

```python
# Read Dataset
file_path = "4876477-Product_Sales_Dataset.xlsx"
df = pd.read_excel(file_path)
```

## Step 4: Exploratory Data Analysis (EDA)
EDA is performed to understand sales distribution, category-wise performance,
and overall trends in the data.

```python
# Category-wise Aggregation
df.groupby("Category")[["Revenue", "Profit", "Quantity"]].sum()
```

## Step 5: Sales Trend Analysis
Visualize how sales change over time.

```python
# Line Plot (Trend)
df_sorted = df.sort_values("Order_Date")

plt.figure()
plt.plot(df_sorted["Order_Date"], df_sorted["Revenue"])
plt.xlabel("Order Date")
plt.ylabel("Revenue")
plt.title("Sales Trend Over Time")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Step 6: Category-wise Sales Comparison

```python
# Bar Chart
plt.figure()
df.groupby("Category")["Revenue"].sum().plot(kind="bar")
plt.xlabel("Category")
plt.ylabel("Total Revenue")
plt.title("Category-wise Sales")
plt.tight_layout()
plt.show()
```

```python
# Count Plot
plt.figure()
sns.countplot(x="Category", data=df)
plt.title("Transaction Distribution by Category")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Key Findings and Conclusion
- The dataset was successfully cleaned and analyzed.
- The Technology and Office Supplies categories generate the highest revenue.
- Sales show noticeable variation across time, indicating seasonal trends.
- Certain sub-categories contribute disproportionately to overall revenue.
- These insights can help organizations optimize inventory and sales strategies.
