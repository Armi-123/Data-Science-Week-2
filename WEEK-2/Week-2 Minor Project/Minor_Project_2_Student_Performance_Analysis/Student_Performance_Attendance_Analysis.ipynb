```python
# Import Required Libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.show()
```

## Step 1: Load the Student Dataset
The dataset is loaded using Pandas from an Excel file stored
in the same folder as this notebook.

```python
# Read Dataset
file_path = "4870116-Student_Academic_Performance.xlsx"
df = pd.read_excel(file_path)
```

## Step 4: Overall Performance Statistics
Analyze the general performance of students using summary statistics.

```python
df.describe()
```

## Step 5: Subject-wise Performance Analysis
Compare student performance across different subjects.

```python
df.groupby("Course")[["GPA", "CGPA"]].mean()
```

## Step 6: Attendance vs Performance Analysis
Analyze how attendance affects student performance.

```python
df.groupby("Absence Days Under-7")[["GPA", "CGPA"]].mean()
```

## Step 7: Distribution of Marks

```python
# Histogram (Marks Distribution)
plt.figure()
plt.hist(df["GPA"], bins=10)
plt.xlabel("GPA")
plt.ylabel("Number of Students")
plt.title("Distribution of Student GPA")
plt.show()
```

## Step 8: Subject-wise Performance Visualization

```python
# Box Plot
plt.figure()
sns.boxplot(x="Course", y="GPA", data=df)
plt.xticks(rotation=45)
plt.title("Course-wise Student Performance")
plt.show()
```

## Step 9: Attendance Category Analysis

```python
# Bar plot
plt.figure()
df["Absence Days Under-7"].value_counts().plot(kind="bar")
plt.xlabel("Attendance Category")
plt.ylabel("Number of Students")
plt.title("Attendance Distribution")
plt.show()
```

## Step 10: Key Findings and Conclusion
- Students with higher attendance generally perform better academically.
- Certain subjects show higher variation in student performance.
- Marks distribution highlights overall performance trends.
- These insights can help institutions improve academic planning and student support.
