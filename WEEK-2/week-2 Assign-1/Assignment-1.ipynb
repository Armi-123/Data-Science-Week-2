## 📊 Supermarket Sales Data Analysis

This project analyzes a supermarket sales dataset using Python, Pandas,
Matplotlib, and Seaborn. The objective is to clean the data, explore sales
patterns, visualize insights, and answer analytical business questions.

## Task 1: Data Loading and Inspection

In this task, we load the dataset and inspect its structure, size, data types,
and missing values.

```python
# import library 
import pandas as pd
```

```python
# Load Dataset
df = pd.read_csv("SuperMarket Analysis.csv")
```

```python
# First 10 Rows
df.head(10)
```

```python
# Rows & Columns
df.shape
```

```python
# Data Types & Structure
df.info()
```

```python
# Missing Values
df.isnull().sum()
```

## Task 2: Data Cleaning and Preparation

This task involves handling missing values, converting the Date column,
and creating a TotalSales column.

```python
# Handle Missing Values
df = df.dropna()
df
```

```python
# Convert Date Column
df['Date'] = pd.to_datetime(df['Date'])
df
```

```python
# Create TotalSales Column
if 'TotalSales' not in df.columns:
    df['TotalSales'] = df['Unit price'] * df['Quantity']

df
```

## Task 3: Exploratory Data Analysis

We analyze product sales, branch performance, and customer spending behavior.

```python
# Top 3 Product Lines by Sales
df.groupby('Product line')['TotalSales'].sum().sort_values(ascending=False).head(3)
```

```python
# Branch with Highest Average Sales
df.groupby('Branch')['TotalSales'].mean().sort_values(ascending=False)
```

```python
# Customer Type Distribution
df['Customer type'].value_counts()
```

```python
# Average Spending by Customer Type
df.groupby('Customer type')['TotalSales'].mean()
```

## Task 4: Visualizations

Visual representations are used to better understand sales trends and patterns.

```python
# Import Libraries
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Bar Chart: Sales by Product Line
df.groupby('Product line')['TotalSales'].sum().plot(kind='bar', figsize=(10,6))
plt.title("Total Sales by Product Line")
plt.show()
```

```python
# Box Plot: Sales by Branch
sns.boxplot(x='Branch', y='TotalSales', data=df)
plt.title("Sales Distribution Across Branches")
plt.show()
```

```python
# Histogram: TotalSales
plt.hist(df['TotalSales'], bins=30)
plt.title("Distribution of Total Sales")
plt.show()
```

```python
# Pairplot
sns.pairplot(df.select_dtypes(include=['int64','float64']))
plt.show()
```

```python
# Correlation Heatmap
sns.heatmap(df.select_dtypes(include=['int64','float64']).corr(),
            annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

## Task 5: Analytical Insight Questions

```python
# Consistent Product Line Across Branches
df.groupby(['Product line','Branch'])['TotalSales'].mean().unstack()
```

```python
# Gender Influence on Sales
df.groupby('Gender')['TotalSales'].mean()
```

## Final Conclusion

- Certain product lines show consistent sales performance across all branches.
- Average spending by Male and Female customers is similar, indicating gender
  does not significantly influence sales value.
- Customer membership plays a role in spending behavior and can be leveraged
  for targeted loyalty programs.

Overall, this analysis demonstrates how data-driven insights can support
better business decisions in sales strategy, inventory planning, and marketing.
